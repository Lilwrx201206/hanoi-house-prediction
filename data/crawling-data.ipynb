```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

```python
data_list=[]



def crawling_data(url):
    response =requests.get(url)
    if response.status_code==200:
        soup =BeautifulSoup(response.content,'html.parser')
        table= soup.find_all('article', class_='property-item')
        for item in table:
            price = item.find('span', class_='price').text.strip()

            area = item.find('span', class_='area').text.strip()

            bedroom_tag = item.find('span', class_='bedroom')
            bedrooms = bedroom_tag.text.strip() if bedroom_tag else "N/A"

            address = item.find('p', class_='new-address').text.strip()

            data_list.append({
            'price': price,
            'area': area,
            'bedrooms': bedrooms,
            'address': address
            })
```

```python
data_list=[]
for i in range(2,22):
    url = f"https://alonhadat.com.vn/nha-dat/can-ban/nha-dat/ha-noi/412/hoan-kiem/trang--{i}.html"
    crawling_data(url)

df = pd.DataFrame(data_list)
file_name = 'hanoi_house_prices_in_hoan_kiem.csv'
df.to_csv(file_name, index=False, encoding='utf-8-sig')
print(f"Tổng số dữ liệu thu thập được: {len(data_list)}")
```
